# PayRoute AI — light EDA

**Synthetic data.** Run `python ml/generate_data.py` first so `data/transactions.csv` exists.

```python
import pandas as pd
df = pd.read_csv('../data/transactions.csv')
print(df.shape)
df.head()
```

```python
# overall failure rate + rate by channel
print('overall failure rate:', round(df['failed'].mean(), 4))
df.groupby('channel')['failed'].mean().sort_values(ascending=False)
```

```python
# failure rate vs latent corridor reliability (should be strongly negative)
import numpy as np
df['rel_bucket'] = pd.cut(df['corridor_reliability'], bins=np.linspace(0, 1, 11))
df.groupby('rel_bucket', observed=True)['failed'].mean()
```

```python
# failure rate by hour of day
df.groupby('hour')['failed'].mean().plot(kind='bar', figsize=(10, 3), title='Failure rate by hour')
```
